# Pilotes et estimation LS
Canal plat sans bruit. Voir expérience 14 pour les canaux sélectifs, le bruit et le bilan énergétique.


```python
import numpy as np

from ofdm_lab.configuration import OFDMConfig
from ofdm_lab.estimation import PilotLayout, estimate_channel_ls, receive_pilots, transmit_pilots
from ofdm_lab.receiver import receive

layout = PilotLayout(OFDMConfig(order=16), spacing=4)
bits = np.random.default_rng(42).integers(0, 2, 100 * layout.bits_per_frame)
received = transmit_pilots(bits, layout).waveform * (0.4 + 0.7j)
estimated = estimate_channel_ls(receive(received, layout.config).grid, layout)
decided = receive_pilots(received, layout, channel_response=estimated, equalizer="zf")
print(f"Pilotes : {layout.pilot_positions.size} ; données : {layout.data_positions.size}")
print(f"Bits utiles : {bits.size} ; erreurs : {np.count_nonzero(decided != bits)}")
```
