# Comparaison appariée OFDM
Démonstration de l’API sur 32 trames ; les validations statistiques sont celles de l’expérience 15.


```python
from ofdm_lab.comparison import simulate_comparison
from ofdm_lab.configuration import OFDMConfig
from ofdm_lab.estimation import PilotLayout

layout = PilotLayout(OFDMConfig(order=16), spacing=4)
rows = simulate_comparison(layout, [1, 0, 0.25j], 12, 32, 42)
for row in rows:
    print(
        f"{row['method']}: {row['bit_errors']}/{row['n_bits']} erreurs, "
        f"MSE symboles={row['symbol_mse']:.6f}"
    )
```
