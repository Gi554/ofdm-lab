# Chaîne OFDM idéale et bilan CP
FFT unitaire, DC/Nyquist nulles, début connu. Aucun canal multitrajet.

```python
import numpy as np

from ofdm_lab import generate_bits
from ofdm_lab.configuration import OFDMConfig
from ofdm_lab.ofdm_simulation import energy_budget
from ofdm_lab.receiver import receive
from ofdm_lab.transmitter import transmit

config = OFDMConfig()
bits = generate_bits(20 * config.bits_per_ofdm_symbol, 20261016)
tx = transmit(bits, config)
assert np.array_equal(receive(tx.waveform, config).bits, bits)
print("Bits restitués :", bits.size)
print("Pénalité CP totale (dB) :", energy_budget(config, 0, "total")["cp_penalty_db"])
```
