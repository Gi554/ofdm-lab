## QAM : énergie moyenne et restitution
L’énergie 1 est une moyenne sur l’alphabet ; un tirage fini peut différer.

```python
from ofdm_lab import demodulate_qam, modulate_qam

for order in [16, 64]:
    k = order.bit_length() - 1
    labels = ((np.arange(order)[:, None] >> np.arange(k - 1, -1, -1)) & 1).astype(np.uint8)
    alphabet = modulate_qam(labels.ravel(), order)
    assert np.array_equal(demodulate_qam(alphabet, order), labels.ravel())
    print(order, "QAM ; Es alphabet =", np.mean(abs(alphabet) ** 2))
```
