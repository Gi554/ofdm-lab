## Comparer des mesures finies à la théorie
Les IC BER sont conservateurs par position binaire pour respecter la dépendance au sein des symboles QAM.

```python
from ofdm_lab.simulation import simulate_awgn

for order in [2, 4, 16, 64]:
    result = simulate_awgn(order, 6, 120000, 20261016 + order)
    print(
        result["modulation"],
        "bits =",
        result["n_bits"],
        "erreurs =",
        result["bit_errors"],
        "BER =",
        result["ber"],
        "référence =",
        result["ber_theory_exact"],
        "IC =",
        (result["ber_ci_lower"], result["ber_ci_upper"]),
    )
```

À haut Eb/N0, zéro erreur n’implique pas un BER théorique nul. Augmenter N améliore la résolution.
Les modèles exacts supposent AWGN, synchronisation parfaite et symboles équiprobables.
