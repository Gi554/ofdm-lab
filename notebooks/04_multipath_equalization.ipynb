# Canal connu : CP, ZF et MMSE
Démonstration courte, indépendante des effectifs des expériences archivées.
Les décisions MMSE utilisent le gain résiduel. La référence de bruit est celle de l’émetteur.

```python
from ofdm_lab.configuration import OFDMConfig
from ofdm_lab.multipath_simulation import simulate_known_channel

config = OFDMConfig(order=16, cp_length=16)
rows = simulate_known_channel(config, [1, 0, 0.5j], 12, 200, seed=42)
for row in rows:
    print(
        row["method"],
        "bits :",
        row["n_bits"],
        "erreurs :",
        row["bit_errors"],
        "BER :",
        row["ber"],
        "MSE :",
        row["mse"],
    )
assert rows[1]["bit_errors"] == rows[2]["bit_errors"]
```
