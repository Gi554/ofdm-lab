# CFO à début connu
Démonstration sans bruit. Voir docs/experiments/13_frequency_offset.md pour les essais bruités et les limites.


```python
import numpy as np

from ofdm_lab.configuration import OFDMConfig
from ofdm_lab.receiver import receive
from ofdm_lab.synchronization import apply_frequency_offset, estimate_cfo_cp
from ofdm_lab.transmitter import transmit

config = OFDMConfig()
bits = np.random.default_rng(42).integers(0, 2, 100 * config.bits_per_ofdm_symbol)
shifted = apply_frequency_offset(transmit(bits, config).waveform, 0.23, config.n_fft)
estimate = estimate_cfo_cp(shifted, config)
corrected = apply_frequency_offset(shifted, -estimate.normalized_offset, config.n_fft)
print(f"CFO estimé : {estimate.normalized_offset:.12f}")
print(f"Erreurs après correction : {np.count_nonzero(receive(corrected, config).bits != bits)}")
```
