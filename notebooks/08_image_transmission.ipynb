```python
from pathlib import Path

import numpy as np

from ofdm_lab.configuration import OFDMConfig
from ofdm_lab.estimation import PilotLayout
from ofdm_lab.image_simulation import simulate_image
from ofdm_lab.image_transport import load_image

root = Path.cwd()
if not (root / "assets").exists():
    root = root.parent
original = load_image(root / "assets/test_pattern.png")
trial = simulate_image(original, PilotLayout(OFDMConfig(order=16)), [1, 0, 0.25j], None, 42)
print(f"Image identique : {np.array_equal(trial.reception.accepted, original)}")
print(f"CRC global valide : {trial.reception.image_crc_valid}")
print(f"Blocs acceptés : {trial.metrics['valid_packets']}/{trial.metrics['transmitted_packets']}")
```
